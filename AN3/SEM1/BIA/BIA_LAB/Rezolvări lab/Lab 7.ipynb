# Exercitiul 2

```python
x = torch.tensor([-10, 10, 8], dtype=torch.float32)
x.requires_grad_(True)
x.grad
sigmoid = 1/(1+torch.exp(-x))
sigmoid.sum().backward()
x.grad
```
